# Check FICO column

```python
import pandas as pd
```

```python
df = pd.read_csv('Customer Loan Data.csv')

print(df.head())
```

## Check FICO Score Distribution

```python
print(df['fico_score'].describe())
```

### Create FICO Buckets

```python
num_buckets = 5

df['rating'] = pd.qcut(df['fico_score'], q=num_buckets, labels=[1,2,3,4,5])
```

```python
print(df[['fico_score', 'rating']].head())
```

## Analyze Default Risk per Bucket

```python
bucket_summary = df.groupby('rating')['default'].agg(['count','sum','mean'])

print(bucket_summary)
```

### Extract Bucket Boundaries

```python
_, bins = pd.qcut(df['fico_score'], q=num_buckets, retbins=True)

print("FICO Score Boundaries:", bins)
```

### Create Rating Function

```python
def assign_rating(fico_score, bins):
    
    for i in range(len(bins)-1):
        if bins[i] <= fico_score <= bins[i+1]:
            return i+1
    
    return None
```

### Test Function

```python
test_score = 700

rating = assign_rating(test_score, bins)

print(f"FICO Score: {test_score}, Rating: {rating}")
```
